### Q1. What is the Filter method in feature selection, and how does it work?

The Filter method is a feature selection technique that selects features based on statistical measures without involving any machine learning model. 
It works by evaluating each feature independently using metrics like correlation, chi-square test, or mutual information. 
Features are ranked based on their scores, and only the top features are selected.

### Q2. How does the Wrapper method differ from the Filter method in feature selection?

The Wrapper method uses a machine learning model to evaluate feature subsets, while the Filter method relies on statistical techniques. 
Wrapper methods try different combinations of features and select the subset that gives the best model performance. 
They are more accurate but computationally expensive compared to Filter methods.

### Q3. What are some common techniques used in Embedded feature selection methods?

Common techniques include:
- Lasso Regression (L1 regularization)
- Ridge Regression (L2 regularization)
- Decision Trees
- Random Forest feature importance
These methods perform feature selection during the model training process.

### Q4. What are some drawbacks of using the Filter method for feature selection?

Drawbacks include:
- Ignores feature interactions
- May select redundant features
- Not optimized for a specific model
- Less accurate compared to Wrapper methods in some cases

### Q5. In which situations would you prefer using the Filter method over the Wrapper method for feature selection?

Filter methods are preferred when:
- Dataset is very large
- Need fast computation
- Preliminary feature selection is required
- Limited computational resources are available

To select features:
1. Compute correlation or mutual information between each feature and churn.
2. Rank features based on importance.
3. Remove low-score features.
4. Eliminate highly correlated redundant features.
5. Select top features for the model.

### Q7. Soccer match prediction - using Embedded method

Steps:
1. Train models like Random Forest or Lasso.
2. Extract feature importance scores.
3. Remove features with low importance.
4. Retrain the model with selected features.
5. Validate performance using cross-validation.

### Q8. House price prediction - using Wrapper method

Steps:
1. Start with a subset of features.
2. Use techniques like forward selection or backward elimination.
3. Train model and evaluate performance.
4. Add/remove features iteratively.
5. Choose subset with best accuracy.
